```python
import pandas as pd
import glob
import os


DOSSIER_CIBLE = r"C:\Users\skand\OneDrive\Documents\Energy_wall\Energy_Wall\data\Data\5"

def auditer_dossier(dossier):
   # On cherche tous les csv (*.csv)
    chemin_recherche = os.path.join(dossier, "*.csv")
    
    # tous les fichiers,  "*.csv" 
    fichiers = glob.glob(chemin_recherche)
    
    print(f"AUDIT DU DOSSIER : {os.path.abspath(dossier)}")
    print(f"Nombre de fichiers CSV trouves : {len(fichiers)}")
    print("="*60)

    fichiers_inutilisables = []

    for fichier in fichiers:
        nom_fichier = os.path.basename(fichier)
        
        try:
          
            df = pd.read_csv(fichier, nrows=3, sep=None, engine='python')
            
            print(f"\nFICHIER : {nom_fichier}")
            print(f"   Shape (Colonnes) : {df.shape[1]}")
            print(f"   Liste Colonnes : {list(df.columns)}")
            print("-" * 30)
            print("   APERCU DES DONNEES :")
            print(df.head(3).to_string(index=False))
            print("="*60)
            
        except Exception as e:
            print(f"\nERREUR sur {nom_fichier} : {e}")
            fichiers_inutilisables.append(nom_fichier)

    # Resume final
    if fichiers_inutilisables:
        print("\nATTENTION : Ces fichiers n'ont pas pu etre lus :")
        for f in fichiers_inutilisables:
            print(f"   - {f}")

if __name__ == "__main__":
    try:
        auditer_dossier(DOSSIER_CIBLE)
    except Exception as e:
        print(f"Erreur generale : {e}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from IPython.display import display, Markdown

# Configuration du style
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (14, 7)
plt.rcParams['font.size'] = 12

def visualiser_science_sante_avec_texte():
    # --- TEXTE D'INTRODUCTION ---
    intro_text = """
# PARTIE 5 : LA DÉPENDANCE VITALE
## "Pourquoi on ne peut pas tout couper ?"

**Résumé Exécutif :**
Monsieur le Ministre, voici le dilemme moral du "Mur Énergétique".
L'IA n'est pas seulement un outil de productivité ou de divertissement. Elle est devenue le poumon de la recherche scientifique mondiale.
Cette section démontre que :
1.  La médecine dépend désormais de l'IA pour les essais cliniques.
2.  L'IA surpasse les médecins humains, offrant un accès aux soins de pointe pour tous.
3.  L'IA (AlphaFold) a résolu en 2 ans des problèmes biologiques vieux de 50 ans.
**Couper l'énergie, c'est aussi couper l'espoir médical.**
    """
    display(Markdown(intro_text))

    print("Génération du Graphique 1 (Essais Cliniques)...")

    # --- GRAPHIQUE 1 : L'IA DANS LES ESSAIS CLINIQUES ---
    path_trials = 'data/Data/5/fig_5.4.19.csv'
    
    if os.path.exists(path_trials):
        df = pd.read_csv(path_trials)
        
        plt.figure()
        sns.lineplot(data=df, x='Year', y='Number of clinical trials', marker='o', linewidth=4, color='green')
        
        plt.title('DÉPENDANCE VITALE : Explosion des essais cliniques utilisant l\'IA', 
                  fontsize=16, fontweight='bold', pad=20)
        plt.xlabel('Année')
        plt.ylabel('Nombre d\'Essais Cliniques')
        
        # Annotation
        try:
            plt.annotate('Accélération exponentielle\nrecherche médicale', 
                         xy=(2023, df['Number of clinical trials'].max()), 
                         xytext=(2019, df['Number of clinical trials'].max()),
                         arrowprops=dict(facecolor='green', shrink=0.05), color='green', fontweight='bold', fontsize=12)
        except: pass

        plt.tight_layout()
        plt.show()

    # --- TEXTE D'ANALYSE 1 ---
    analyse_1 = """
### 1. L'Accélération de la Recherche Médicale
Regardez cette courbe. Ce n'est pas de la spéculation financière, ce sont des médicaments en cours de test.
* **Le constat :** Le nombre d'essais cliniques optimisés par l'IA explose.
* **L'enjeu :** L'IA permet de réduire le temps de développement d'un médicament de 10 ans à 2 ans. C'est une question de vie ou de mort pour les patients.
    """
    display(Markdown(analyse_1))

    print("Génération du Graphique 2 (Précision)...")

    # --- GRAPHIQUE 2 : L'INTELLIGENCE MÉDICALE ---
    path_med_acc = 'data/Data/5/fig_5.4.1.csv'
    
    if os.path.exists(path_med_acc):
        df = pd.read_csv(path_med_acc)
        # Nettoyage du pourcentage
        df['Accuracy'] = df['Test accuracy'].astype(str).str.rstrip('%').astype(float)
        
        # Tri par année pour voir la progression
        df = df.sort_values('Year')
        
        plt.figure()
        sns.barplot(data=df, x='Model', y='Accuracy', palette='Blues')
        
        plt.title('PERFORMANCE : En 3 ans, l\'IA est devenue un expert médical (De 38% à 90%+)', 
                  fontsize=16, fontweight='bold', pad=20)
        plt.xlabel('Modèle (Année)')
        plt.ylabel('Précision au Test Médical (%)')
        plt.ylim(0, 100)
        
        # Seuil critique
        plt.axhline(60, color='red', linestyle='--', label='Seuil de réussite examen médecin (USMLE)')
        plt.text(0, 62, "Seuil de réussite Médecin", color='red', fontweight='bold')
        
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

    # --- TEXTE D'ANALYSE 2 ---
    analyse_2 = """
### 2. Le Super-Médecin Numérique
L'argument de la "futilité" de l'IA tombe ici.
* **Le fait :** En 2021, l'IA échouait aux examens de médecine (38%). Aujourd'hui, elle les réussit mieux que la plupart des humains (90%+).
* **L'opportunité :** C'est la solution aux déserts médicaux. Un diagnostic de niveau expert accessible partout, pour peu qu'on ait l'énergie pour le faire tourner.
    """
    display(Markdown(analyse_2))

    print("Génération du Graphique 3 (AlphaFold)...")

    # --- GRAPHIQUE 3 : LA RÉVOLUTION BIOLOGIQUE ---
    path_bio = 'data/Data/5/fig_5.2.4.csv'
    
    if os.path.exists(path_bio):
        df = pd.read_csv(path_bio)
        
        # On se concentre sur AlphaFold (la révolution IA) vs PDB (méthode classique)
        df = df[df['Label'].isin(['AlphaFold DB', 'PDB'])]
        
        plt.figure()
        sns.lineplot(data=df, x='Year', y='Number of entities (log scale)', hue='Label', marker='o', linewidth=3)
        
        plt.title('RÉVOLUTION SCIENTIFIQUE : L\'IA (AlphaFold) a résolu la structure des protéines', 
                  fontsize=16, fontweight='bold', pad=20)
        plt.xlabel('Année')
        plt.ylabel('Nombre de Structures (Échelle Log)')
        plt.yscale('log')
        
        plt.text(2023, 100000000, "200+ Millions de structures\n(Explosion de la connaissance)", 
                 color='orange', fontweight='bold', ha='center')

        plt.tight_layout()
        plt.show()

    # --- TEXTE DE CONCLUSION ---
    conclusion = """
### 3. Verdict : L'Usage "Noble" de l'Énergie
Le dernier graphique (AlphaFold) est historique. En orange (méthode classique), la science avançait pas à pas. En bleu (IA), elle a fait un bond de géant, cartographiant tout le vivant en 18 mois.
**Recommandation Finale :** Le problème n'est pas la consommation de l'IA en soi, mais son usage. Nous devons réserver nos ressources énergétiques (et nos minéraux) à ces usages vitaux (Santé, Science) et réguler les usages futiles.
    """
    display(Markdown(conclusion))

if __name__ == "__main__":
    visualiser_science_sante_avec_texte()
```
