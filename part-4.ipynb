```python
import pandas as pd
import glob
import os


DOSSIER_CIBLE = r"C:\Users\skand\OneDrive\Documents\Energy_wall\Energy_Wall\data\Data\4"

def auditer_dossier(dossier):
   # On cherche tous les csv (*.csv)
    chemin_recherche = os.path.join(dossier, "*.csv")
    
    # tous les fichiers,  "*.csv" 
    fichiers = glob.glob(chemin_recherche)
    
    print(f"AUDIT DU DOSSIER : {os.path.abspath(dossier)}")
    print(f"Nombre de fichiers CSV trouves : {len(fichiers)}")
    print("="*60)

    fichiers_inutilisables = []

    for fichier in fichiers:
        nom_fichier = os.path.basename(fichier)
        
        try:
          
            df = pd.read_csv(fichier, nrows=3, sep=None, engine='python')
            
            print(f"\nFICHIER : {nom_fichier}")
            print(f"   Shape (Colonnes) : {df.shape[1]}")
            print(f"   Liste Colonnes : {list(df.columns)}")
            print("-" * 30)
            print("   APERCU DES DONNEES :")
            print(df.head(3).to_string(index=False))
            print("="*60)
            
        except Exception as e:
            print(f"\nERREUR sur {nom_fichier} : {e}")
            fichiers_inutilisables.append(nom_fichier)

    # Resume final
    if fichiers_inutilisables:
        print("\nATTENTION : Ces fichiers n'ont pas pu etre lus :")
        for f in fichiers_inutilisables:
            print(f"   - {f}")

if __name__ == "__main__":
    try:
        auditer_dossier(DOSSIER_CIBLE)
    except Exception as e:
        print(f"Erreur generale : {e}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from IPython.display import display, Markdown

# Configuration du style
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (14, 7)
plt.rcParams['font.size'] = 12

def visualiser_economie_ia_avec_texte():
    # --- TEXTE D'INTRODUCTION ---
    intro_text = """
# PARTIE 4 : L'ADDICTION ÉCONOMIQUE
## "Pourquoi le retour en arrière est impossible ?"

**Résumé Exécutif :**
Monsieur le Ministre, l'argument le plus fort contre une régulation brutale n'est pas technologique, il est économique.
L'IA n'est plus un projet de laboratoire. Elle est devenue le moteur de la croissance mondiale, des ressources humaines et des investissements.
Cette section prouve que nous sommes dans une situation de **"Too Big to Fail"** (Trop gros pour échouer).
    """
    display(Markdown(intro_text))

    print("Génération du Graphique 1 (Emploi)...")

    # --- GRAPHIQUE 1 : LA RUÉE VERS L'OR (Offres d'emploi) ---
    path_skills = 'data/Data/4/fig_4.2.5.csv'
    
    if os.path.exists(path_skills):
        df = pd.read_csv(path_skills)
        
        # On ne garde que les compétences clés pour la lisibilité
        df_subset = df[df['Generative AI skill'].isin(['Generative artificial intelligence', 'Large language modeling', 'Prompt engineering'])]
        
        plt.figure()
        sns.barplot(data=df_subset, x='Number of AI Job Postings', y='Generative AI skill', palette='magma')
        
        plt.title('DÉPENDANCE RH : L\'Explosion des recrutements en IA Générative (2023)', 
                  fontsize=16, fontweight='bold', pad=20)
        plt.xlabel('Nombre d\'Offres d\'Emploi')
        plt.ylabel('')
        
        # Annotation explicative
        plt.text(10000, 0, "Compétence N°1 :\nIndispensable pour les entreprises", 
                 color='white', fontweight='bold', ha='center')

        plt.tight_layout()
        plt.show()

    # --- TEXTE D'ANALYSE 1 ---
    analyse_1 = """
### 1. La Transformation du Marché du Travail
L'économie réelle a basculé.
* **Le fait :** En 2023, les compétences liées à l'IA Générative (Prompt Engineering, LLM) ont explosé dans les offres d'emploi.
* **L'implication :** Des milliers d'emplois dépendent désormais de la disponibilité de ces outils. Couper l'énergie des serveurs, c'est mettre une partie de la main-d'œuvre qualifiée au chômage technique.
    """
    display(Markdown(analyse_1))

    print("Génération du Graphique 2 (Adoption)...")

    # --- GRAPHIQUE 2 : L'ADOPTION FOUDROYANTE ---
    path_adoption = 'data/Data/4/fig_4.4.1.csv'
    
    if os.path.exists(path_adoption):
        df = pd.read_csv(path_adoption)
        # Nettoyage des pourcentages
        df['Pct'] = df['% of respondents'].astype(str).str.rstrip('%').astype(float)
        
        plt.figure()
        # Courbe d'adoption
        sns.lineplot(data=df, x='Year', y='Pct', hue='Label', marker='o', linewidth=4)
        
        plt.title('IRRÉVERSIBILITÉ : L\'Adoption de la GenAI a doublé en un an (33% -> 71%)', 
                  fontsize=16, fontweight='bold', pad=20)
        plt.ylabel('Taux d\'Adoption (%)')
        plt.xlabel('Année')
        plt.ylim(0, 100)
        
        # Flèche explicative
        plt.annotate('Adoption Massive\n= Dépendance Totale', 
                     xy=(2024, 71), xytext=(2023.5, 90),
                     arrowprops=dict(facecolor='red', shrink=0.05), color='red', fontweight='bold', fontsize=12)

        plt.tight_layout()
        plt.show()

    # --- TEXTE D'ANALYSE 2 ---
    analyse_2 = """
### 2. Le Point de Non-Retour
La vitesse d'adoption est sans précédent dans l'histoire technologique (plus rapide que le smartphone ou internet).
* **Le chiffre clé :** En seulement 12 mois, nous sommes passés de 33% à **71% d'entreprises** utilisant l'IA.
* **Le constat :** L'IA est désormais intégrée dans les processus métiers (RH, Marketing, R&D). Revenir en arrière n'est plus une option opérationnelle pour nos entreprises.
    """
    display(Markdown(analyse_2))

    print("Génération du Graphique 3 (Investissement)...")

    # --- GRAPHIQUE 3 : LES MILLIARDS DE L'IA ---
    path_invest = 'data/Data/4/fig_4.3.2.csv'
    
    if os.path.exists(path_invest):
        df = pd.read_csv(path_invest)
        
        plt.figure()
        sns.barplot(data=df, x='Year', y='Total investment (in billions of US dollars)', color='navy')
        
        plt.title('ENJEU FINANCIER : Les investissements mondiaux explosent', 
                  fontsize=16, fontweight='bold', pad=20)
        plt.ylabel('Investissements (Milliards $)')
        plt.xlabel('Année')
        
        plt.tight_layout()
        plt.show()

    # --- TEXTE DE CONCLUSION ---
    conclusion = """
### 3. Verdict : Une Bulle trop grosse pour éclater ?
Le dernier graphique montre des centaines de milliards de dollars d'investissements.
Ce n'est pas de la spéculation virtuelle : c'est de l'infrastructure lourde (Usines de puces, Data Centers).
**Conclusion Politique :** Nous ne pouvons pas arrêter cette machine économique. La seule voie viable est de **l'orienter** vers des modèles sobres (Partie 2) et utiles (Partie 5), plutôt que de tenter de la freiner.
    """
    display(Markdown(conclusion))

if __name__ == "__main__":
    visualiser_economie_ia_avec_texte()
```
