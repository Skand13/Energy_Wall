```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from IPython.display import display, Markdown

# Configuration du style
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11
plt.rcParams['axes.titlesize'] = 16
plt.rcParams['axes.titleweight'] = 'bold'
plt.rcParams['lines.linewidth'] = 3

def print_header(title, subtitle=""):
    display(Markdown(f"# {title}"))
    if subtitle:
        display(Markdown(f"_{subtitle}_"))
    display(Markdown("---"))

# 0. INTRODUCTION
print_header("DOSSIER : LE MUR ÉNERGÉTIQUE DE L'IA", "Introduction & Contexte")

intro_text = """
**Le Constat :** Ce matin, un incident a eu lieu (train arrêté, panne). Ce n'est pas un hasard, c'est un symptôme.
L'infrastructure physique (réseau électrique, eau, métaux) ne suit plus la cadence imposée par la révolution numérique.

**Notre Feuille de Route :**
1. **Le Mur (Apocalyps) :** Pourquoi nous allons dans l'impasse.
2. **La Solution (Bifurcation) :** La technologie pour nous sauver (SLM).
3. **L'Action (Politique) :** Comment financer cette transition.
"""
display(Markdown(intro_text))

# I. PARTIE APOCALYPS
print_header("I. APOCALYPS : LA COURSE À L'ABÎME", "Le Mur Physique")

display(Markdown("""
**1. Le Piège des Lois d'Échelle (Scaling Laws) :**
Pour gagner un peu d'intelligence, il faut multiplier la consommation par 10 ou 100.
"""))

# Données pour Graphique 1 (CO2)
years = [2012, 2014, 2016, 2018, 2020, 2022, 2024]
co2_emissions = [1, 5, 25, 200, 1500, 6000, 18000] 

# Données pour Graphique 2 (Lithium)
years_min = [2020, 2025, 2030, 2035]
demand_lithium = [100, 350, 900, 2000] 

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Plot CO2
ax1.plot(years, co2_emissions, color='darkred', marker='o')
ax1.set_title("Explosion des Émissions CO2 (Entraînement)")
ax1.set_ylabel("Émissions (Tonnes Éq. CO2)")
ax1.annotate('AlexNet (2012)', xy=(2012, 1), xytext=(2013, 2000), arrowprops=dict(arrowstyle='->'))
ax1.annotate('GPT-4 / Gemini', xy=(2024, 18000), xytext=(2020, 15000), 
             arrowprops=dict(facecolor='black', shrink=0.05))
ax1.text(2018, 8000, "LE MUR\nÉNERGÉTIQUE", color='red', fontsize=14, fontweight='bold')
ax1.grid(True, alpha=0.3)

# Plot Lithium
ax2.bar(years_min, demand_lithium, color='purple', alpha=0.7, width=3)
ax2.plot(years_min, demand_lithium, color='black', linestyle='--')
ax2.set_title("Le Mur des Ressources (Demande Lithium)")
ax2.set_ylabel("Demande (Base 100 = 2020)")
ax2.text(2030, 1000, "+30% / an\n(Insoutenable)", ha='center', color='purple', fontweight='bold')

plt.tight_layout()
plt.show()

display(Markdown("""
**Verdict :** Si on continue la courbe rouge, on manque de métal (courbe violette) et d'électricité avant 2030.
"""))

# II. LA BIFURCATION (PHI-3 & SLM)
print_header("II. LA SOLUTION TECHNIQUE", "Faire mieux avec moins")

display(Markdown("""
**Le Mythe :** \"Plus c'est gros, mieux c'est.\"
**La Réalité :** Des petits modèles (SLM) comme Phi-3 égalent les géants pour 100x moins d'énergie.
"""))

# Données Graphique 3
models = ['Géant (GPT-4)', 'Moyen (Llama-2)', 'Petit (Phi-3)']
params = [1000, 70, 3.8]  
perf = [90, 70, 85]       

fig, ax1 = plt.subplots(figsize=(10, 5))

# Barres performance
bars = ax1.bar(models, perf, color=['lightgrey', 'lightgrey', 'green'], alpha=0.8, label='Intelligence (Score)')
ax1.set_ylabel('Score de Performance (0-100)', color='green', fontweight='bold')
ax1.set_ylim(0, 100)
ax1.bar_label(bars, fmt='%.0f pts', padding=3)

# Ligne consommation
ax2 = ax1.twinx()
ax2.plot(models, params, color='red', marker='D', markersize=10, linewidth=2, linestyle='--', label='Consommation (Taille)')
ax2.set_ylabel('Taille / Conso (Log Scale)', color='red', fontweight='bold')
ax2.set_yscale('log') 

plt.title("Rupture Technologique : Phi-3 (Vert) casse la corrélation Taille/Performance")
plt.show()

# III. RÉALISME ÉCONOMIQUE
print_header("III. LE RÉALISME ÉCONOMIQUE", "Le Marché est Prêt")

# Données Graphique 4
years_inv = [2018, 2019, 2020, 2021, 2022, 2023]
investments = [40, 55, 70, 120, 100, 160] 

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 5))

# Plot Investissements
ax1.fill_between(years_inv, investments, color='gold', alpha=0.3)
ax1.plot(years_inv, investments, color='orange', linewidth=3)
ax1.set_title("La Bulle Financière (Investissements IA)")
ax1.set_ylabel("Milliards USD")
ax1.text(2021, 130, "Inertie Financière\n(Too big to fail)", fontweight='bold')

# Plot Demande RSE
labels = ['Considèrent le Coût Carbone\ncomme CRITIQUE', 'Ne s\'en soucient pas']
sizes = [30, 70]
colors = ['#4CAF50', '#E0E0E0'] 
ax2.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90, explode=(0.1, 0))
ax2.set_title("Demande du Marché (Enquête Entreprises)")

plt.tight_layout()
plt.show()

display(Markdown("**Conclusion :** Il y a trop d'argent engagé pour tout arrêter (Graphique Gauche), mais les clients veulent de la sobriété (Graphique Droite). C'est l'opportunité."))

# IV. UTILITÉ SOCIALE
print_header("IV. UTILITÉ SOCIALE", "Réserver l'énergie pour le vital")

display(Markdown("""
**L'exemple AlphaFold :** L'IA n'est pas qu'un chatbot. Elle a résolu en 18 mois un problème biologique vieux de 50 ans.
Il faut réserver l'énergie pour **ça**.
"""))

# Données Graphique 5
categories = ['Méthode Classique\n(Humain seul)', 'IA (AlphaFold)']
speed = [1, 50] 

plt.figure(figsize=(8, 4))
bars = plt.barh(categories, speed, color=['grey', 'cyan'])
plt.title("Vitesse de Découverte Scientifique (Biologie)")
plt.xlabel("Facteur d'accélération (x fois plus vite)")
plt.text(45, 1, "Révolution", color='blue', fontweight='bold', va='center')

plt.tight_layout()
plt.show()

# V. CONCLUSION & PLAN POLITIQUE
print_header("V. PLAN D'ACTION POLITIQUE", "Conclusion")

final_plan = """
Monsieur le Ministre, pour éviter le "Mur" tout en gardant le leadership technologique :

1.  **FLÉCHER L'INVESTISSEMENT :**
    * Stop aux subventions pour les Data Centers "bruts".
    * **Oui** aux subventions pour l'optimisation (SLM / Puces Neuromorphiques).

2.  **ÉCO-CONDITIONNALITÉ :**
    * Accès prioritaire au réseau électrique pour les projets d'IA à utilité publique (Santé/Recherche).
    * Taxe carbone sur les modèles génératifs purement ludiques ou inefficaces.

3.  **TRANSPARENCE :**
    * Obliger les GAFAM à publier le ratio "Watt par Token" de leurs modèles.

**L'IA ne doit pas être une facture énergétique, mais un investissement précis.**
"""
display(Markdown(final_plan))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from IPython.display import display, Markdown

# Configuration du style graphique
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (14, 8)
plt.rcParams['font.size'] = 11
plt.rcParams['axes.titlesize'] = 15
plt.rcParams['axes.titleweight'] = 'bold'
plt.rcParams['lines.linewidth'] = 3

def afficher_titre_section(titre, sous_titre=""):
    display(Markdown("---"))
    display(Markdown(f"# {titre}"))
    if sous_titre:
        display(Markdown(f"**{sous_titre}**"))
    display(Markdown("---"))

# INTRODUCTION
afficher_titre_section("INTRODUCTION : LE BLOCAGE (2029)", "L'incident du train et le contexte")

intro_text = """
Nous sommes le lundi 12 décembre 2029. Votre train s'arrête en pleine campagne. Ce n'est pas une panne isolée, c'est un délestage réseau.
L'infrastructure électrique européenne ne tient plus la charge.

Pourquoi ? Parce que nous avons construit une économie numérique exponentielle sur une infrastructure physique linéaire.
Ce dossier APOCALYPS ne parle pas du futur lointain, mais des courbes mathématiques qui se croisent aujourd'hui.
"""
display(Markdown(intro_text))

# PARTIE 1 : LA MECANIQUE DU CRASH
afficher_titre_section("I. LA MÉCANIQUE DU CRASH", "Pourquoi l'infrastructure ne peut plus suivre")

display(Markdown("""
**Le Moteur de l'Apocalypse : La fin de la Loi de Moore**
Jusqu'à récemment, les ordinateurs devenaient plus efficaces aussi vite qu'ils devenaient plus puissants (Loi de Moore). C'est fini.
Aujourd'hui, la demande de calcul de l'IA (Ligne Rouge) explose verticalement, tandis que le matériel (Ligne Grise) stagne.
Le seul moyen de combler l'écart est de consommer plus d'énergie brute.
"""))

# Graphique 1 : Le Décrochage (Compute vs Moore)
years_compute = [2012, 2014, 2016, 2018, 2020, 2022, 2024]
# Echelle Logarithmique simulée pour la visualisation
ai_demand = [1, 10, 100, 1000, 10000, 100000, 1000000] 
hardware_progress = [1, 2, 4, 8, 16, 32, 64] 

plt.figure(figsize=(12, 6))
plt.plot(years_compute, ai_demand, color='#d62728', marker='o', label='Demande Calcul IA (x10/an)')
plt.plot(years_compute, hardware_progress, color='grey', linestyle='--', label='Progrès Matériel (Loi de Moore)')

plt.yscale('log')
plt.title("LA RUPTURE : La demande IA va plus vite que la physique")
plt.ylabel("Puissance requise (Échelle Log)")
plt.legend()
plt.grid(True, which="both", ls="-", alpha=0.2)

# Annotation du gap
plt.annotate('Le GAP ÉNERGÉTIQUE\n(Comblé par des centrales à charbon/gaz)', 
             xy=(2023, 100000), xytext=(2018, 500000),
             arrowprops=dict(facecolor='black', shrink=0.05), color='red', fontweight='bold')

plt.show()

# PARTIE 2 : LE MUR DE L'EAU ET DE L'ENERGIE
afficher_titre_section("II. IMPACT PHYSIQUE IMMÉDIAT", "Eau et Électricité")

display(Markdown("""
**L'Impact Concret :**
Les Data Centers ne sont pas des nuages virtuels. Ce sont des usines qui consomment de l'eau potable pour se refroidir et de l'électricité pour tourner.
Les chiffres ci-dessous (Source : AIE & RTE) montrent une saturation locale et une consommation unitaire qui explose.
"""))

# Graphique 2 : Eau et Électricité (Subplots)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Données Eau (ChatGPT vs Google)
items = ['Recherche Google', 'Conversation ChatGPT']
values_water = [0.3, 3.0] # Wh (Proxy énergie/eau)

ax1.bar(items, values_water, color=['green', 'red'], alpha=0.7)
ax1.set_title("Le Choc Unitaire (Facteur x10)")
ax1.set_ylabel("Consommation relative")
ax1.text(1, 3.1, "x10 par requête", ha='center', fontweight='bold', color='red')

# Données Saturation Réseau (Exemple Irlande/France Data Centers)
years_grid = ['2022', '2026 (Projection)']
grid_load = [17, 32] # % de la consommation nationale (Exemple Irlande, très parlant)

ax2.fill_between(years_grid, grid_load, color='orange', alpha=0.3)
ax2.plot(years_grid, grid_load, color='darkorange', marker='o')
ax2.set_title("Saturation du Réseau (Part des Data Centers dans la conso nationale)")
ax2.set_ylabel("% de l'électricité du pays")
ax2.set_ylim(0, 40)
ax2.text(1, 33, "Le réseau risque de céder", ha='center', fontweight='bold', color='darkorange')

plt.show()

# PARTIE 3 : LE MUR DES RESSOURCES (MINERAUX)
afficher_titre_section("III. LE MUR GÉOLOGIQUE", "Il n'y a pas assez de mines")

display(Markdown("""
**L'Impasse Matérielle :**
Même si nous produisons assez d'électricité, nous n'avons pas assez de matériaux pour construire les infrastructures.
La demande en Lithium (pour les batteries de secours des Data Centers) et en Cuivre (pour les câbles) suit une courbe exponentielle que l'industrie minière ne peut pas suivre.
"""))

# Graphique 3 : Pénurie de Cuivre et Lithium
minerals = ['Cuivre (Câbles)', 'Lithium (Batteries)']
demand_growth = [20, 300] # % Augmentation d'ici 2035
supply_growth = [5, 20]   # % Augmentation offre minière possible

x = np.arange(len(minerals))
width = 0.35

plt.figure(figsize=(10, 6))
rects1 = plt.bar(x - width/2, demand_growth, width, label='Demande (Besoins IA)', color='purple')
rects2 = plt.bar(x + width/2, supply_growth, width, label='Offre (Capacité Minière)', color='grey')

plt.ylabel('% de Croissance (2024-2035)')
plt.title('LE DÉFICIT STRUCTUREL : La géologie ne suit pas')
plt.xticks(x, minerals)
plt.legend()

# Annotation Pénurie
plt.text(1, 150, "DÉFICIT MASSIF\n(Impossible à combler)", ha='center', color='purple', fontweight='bold', bbox=dict(facecolor='white', edgecolor='purple'))

plt.show()

# CONCLUSION DE LA PARTIE APOCALYPS
afficher_titre_section("CONCLUSION DU DIAGNOSTIC", "L'impasse est physique, pas logicielle")

conclusion_text = """
**Résumé du Mur :**
1.  **Mur de Calcul :** La demande va plus vite que la technologie (Fin de Moore).
2.  **Mur Énergétique :** Le réseau sature déjà (Panne de train).
3.  **Mur Géologique :** Nous ne pouvons pas extraire assez de métal pour tout construire.

Si nous continuons le "Business as Usual", l'effondrement partiel de l'infrastructure est mathématique.
Il faut changer de modèle (ce que nous verrons dans la partie Solutions).
"""
display(Markdown(conclusion_text))
```
