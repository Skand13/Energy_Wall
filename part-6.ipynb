```python
import pandas as pd
import glob
import os


DOSSIER_CIBLE = r"C:\Users\skand\OneDrive\Documents\Energy_wall\Energy_Wall\data\Data\6"

def auditer_dossier(dossier):
   # On cherche tous les csv (*.csv)
    chemin_recherche = os.path.join(dossier, "*.csv")
    
    # tous les fichiers,  "*.csv" 
    fichiers = glob.glob(chemin_recherche)
    
    print(f"AUDIT DU DOSSIER : {os.path.abspath(dossier)}")
    print(f"Nombre de fichiers CSV trouves : {len(fichiers)}")
    print("="*60)

    fichiers_inutilisables = []

    for fichier in fichiers:
        nom_fichier = os.path.basename(fichier)
        
        try:
          
            df = pd.read_csv(fichier, nrows=3, sep=None, engine='python')
            
            print(f"\nFICHIER : {nom_fichier}")
            print(f"   Shape (Colonnes) : {df.shape[1]}")
            print(f"   Liste Colonnes : {list(df.columns)}")
            print("-" * 30)
            print("   APERCU DES DONNEES :")
            print(df.head(3).to_string(index=False))
            print("="*60)
            
        except Exception as e:
            print(f"\nERREUR sur {nom_fichier} : {e}")
            fichiers_inutilisables.append(nom_fichier)

    # Resume final
    if fichiers_inutilisables:
        print("\nATTENTION : Ces fichiers n'ont pas pu etre lus :")
        for f in fichiers_inutilisables:
            print(f"   - {f}")

if __name__ == "__main__":
    try:
        auditer_dossier(DOSSIER_CIBLE)
    except Exception as e:
        print(f"Erreur generale : {e}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from IPython.display import display, Markdown

# Configuration du style
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (14, 7)
plt.rcParams['font.size'] = 12

def visualiser_politique_avec_texte():
    # --- TEXTE D'INTRODUCTION ---
    intro_text = """
# PARTIE 6 : LA RÉPONSE POLITIQUE
## "Qui gagnera la course ?"

**Résumé Exécutif :**
Monsieur le Ministre, la technologie n'évolue pas dans un vide politique.
Face au "Mur Énergétique", les États ne restent pas passifs. Nous assistons à une course aux armements numérique où l'argent public coule à flots.

Cette section analyse la réaction des puissances mondiales :
1.  **L'Argent :** Qui finance la course ? (Domination US).
2.  **La Loi :** Le réveil législatif face aux risques.
3.  **L'Enjeu :** Ne pas devenir une colonie numérique faute d'investissements ciblés (notamment dans les modèles sobres).
    """
    display(Markdown(intro_text))

    print("Génération du Graphique 1 (Dépenses)...")

    # --- GRAPHIQUE 1 : QUI DÉPENSE LE PLUS ? (fig_6.3.1.csv) ---
    path_money = "data/Data/6/fig_6.3.1.csv"
    
    if os.path.exists(path_money):
        try:
            # On lit le fichier des dépenses par pays
            df_money = pd.read_csv(path_money)
            
            # On trie pour avoir les plus gros en premier et on prend le Top 10
            df_money = df_money.sort_values('Tot. Value (M$)', ascending=False).head(10)

            plt.figure()
            barplot = sns.barplot(data=df_money, x='Tot. Value (M$)', y='Country', palette='viridis')
            
            plt.title('GÉOPOLITIQUE : La Course aux Armements IA (Dépenses Publiques)', 
                      fontsize=16, fontweight='bold', pad=20)
            plt.xlabel('Investissement Public (Millions $)')
            plt.ylabel('')
            
            # Annotation pour la domination US
            try:
                valeur_us = df_money.loc[df_money['Country'] == 'United States', 'Tot. Value (M$)'].values[0]
                plt.text(valeur_us, 0, f" USA: {valeur_us:,.0f} M$ \n(Domination Totale)", 
                         va='center', ha='left', fontweight='bold', color='darkblue', fontsize=12)
            except: pass
            
            plt.tight_layout()
            plt.show()

        except Exception as e:
            print(f"Erreur Graphique Dépenses: {e}")

    # --- TEXTE D'ANALYSE 1 ---
    analyse_1 = """
### 1. La Disparité des Moyens
Le graphique est sans appel.
* **Le constat :** Les États-Unis investissent massivement plus que le reste du monde combiné.
* **Le risque :** Si l'Europe (et la France) ne cible pas ses investissements intelligemment (sur l'efficience énergétique et la santé, comme vu en Parties 2 et 5), nous ne pourrons pas rivaliser sur la "force brute".
    """
    display(Markdown(analyse_1))

    print("Génération du Graphique 2 (Législation)...")

    # --- GRAPHIQUE 2 : LA PRISE DE CONSCIENCE POLITIQUE (fig_6.2.14.csv) ---
    path_laws = "data/Data/6/fig_6.2.14.csv"
    
    if os.path.exists(path_laws):
        try:
            # On lit l'évolution des mentions dans les lois
            df_laws = pd.read_csv(path_laws)
            
            plt.figure()
            sns.lineplot(data=df_laws, x='Year', y='Number of mentions', marker='o', linewidth=4, color='crimson')
            
            plt.title("LÉGISLATION : L'Urgence Politique (Mentions de l'IA dans les lois)", 
                      fontsize=16, fontweight='bold', pad=20)
            plt.ylabel("Nombre de Mentions Législatives")
            plt.xlabel("Année")
            
            # Zone de "Réveil"
            plt.axvspan(2019, 2023, color='yellow', alpha=0.2)
            plt.text(2021, df_laws['Number of mentions'].max()*0.8, "Le Réveil Politique\n(Régulation)", 
                     ha='center', fontweight='bold', color='darkorange')
            
            plt.grid(True, linestyle='--')
            plt.tight_layout()
            plt.show()

        except Exception as e:
            print(f"Erreur Graphique Lois: {e}")

    # --- TEXTE DE CONCLUSION ---
    conclusion = """
### 2. Le Réveil du Législateur
Longtemps ignorée, l'IA est devenue un sujet politique majeur depuis 2019.
* **La courbe :** L'explosion des mentions législatives montre que les gouvernements tentent de reprendre le contrôle (AI Act, décrets...).
* **L'opportunité :** C'est le moment idéal pour introduire des critères d'**Éco-Conditionnalité**.
    
---
# CONCLUSION GÉNÉRALE DU DOSSIER "ENERGY WALL"

Monsieur le Ministre, nous avons vu que :
1.  **Le Mur est réel** (Explosion exponentielle de la conso & pénurie de Cuivre/Lithium).
2.  **L'Économie est piégée** (Dépendance totale des entreprises et de l'emploi).
3.  **La Santé est dépendante** (Progrès médicaux vitaux).

**La seule voie de sortie n'est pas l'arrêt, mais la bifurcation.**
Nous devons rediriger l'argent public (Partie 6) vers les modèles sobres (Partie 2) et les usages vitaux (Partie 5), et taxer les modèles énergivores et futiles.
    """
    display(Markdown(conclusion))

if __name__ == "__main__":
    visualiser_politique_avec_texte()
```
