```python
import pandas as pd
import glob
import os


DOSSIER_CIBLE = r"C:\Users\skand\OneDrive\Documents\Energy_wall\Energy_Wall\data\Data2\Data"

def auditer_dossier(dossier):
   # On cherche tous les csv (*.csv)
    chemin_recherche = os.path.join(dossier, "*.csv")
    
    # tous les fichiers,  "*.csv" 
    fichiers = glob.glob(chemin_recherche)
    
    print(f"AUDIT DU DOSSIER : {os.path.abspath(dossier)}")
    print(f"Nombre de fichiers CSV trouves : {len(fichiers)}")
    print("="*60)

    fichiers_inutilisables = []

    for fichier in fichiers:
        nom_fichier = os.path.basename(fichier)
        
        try:
          
            df = pd.read_csv(fichier, nrows=3, sep=None, engine='python')
            
            print(f"\nFICHIER : {nom_fichier}")
            print(f"   Shape (Colonnes) : {df.shape[1]}")
            print(f"   Liste Colonnes : {list(df.columns)}")
            print("-" * 30)
            print("   APERCU DES DONNEES :")
            print(df.head(3).to_string(index=False))
            print("="*60)
            
        except Exception as e:
            print(f"\nERREUR sur {nom_fichier} : {e}")
            fichiers_inutilisables.append(nom_fichier)

    # Resume final
    if fichiers_inutilisables:
        print("\nATTENTION : Ces fichiers n'ont pas pu etre lus :")
        for f in fichiers_inutilisables:
            print(f"   - {f}")

if __name__ == "__main__":
    try:
        auditer_dossier(DOSSIER_CIBLE)
    except Exception as e:
        print(f"Erreur generale : {e}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from IPython.display import display, Markdown

# Configuration du style
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (14, 7)
plt.rcParams['font.size'] = 12

def visualiser_explosion_ia_avec_texte():
    # --- TEXTE D'INTRODUCTION ---
    intro_text = """
# PARTIE 1 : L'EXPLOSION INCONTRÔLÉE
## "Comment en est-on arrivé là ?"

**Résumé Exécutif :**
Nous sommes face à une convergence de courbes exponentielles unique dans l'histoire industrielle. La recherche (logiciel), la puissance de calcul (matériel) et la consommation de ressources (physique) s'emballent simultanément.

Cette section analyse la mécanique de ce crash annoncé en 3 temps :
1.  L'emballement scientifique (Le Tsunami).
2.  La mécanique de l'exponentielle (Loi de Moore dépassée).
3.  La conséquence environnementale directe (CO2).
    """
    display(Markdown(intro_text))

    print("Génération du Graphique 1 (Publications)...")

    # --- GRAPHIQUE 1 : LE TSUNAMI SCIENTIFIQUE ---
    path_pubs = 'data/Data2/Data/fig_1.1.1.csv'
    
    if os.path.exists(path_pubs):
        df = pd.read_csv(path_pubs)
        
        plt.figure()
        sns.lineplot(data=df, x='Year', y='Number of AI publications in CS (in thousands)', marker='o', linewidth=4, color='purple')
        
        plt.title('ACCÉLÉRATION : L\'explosion de la recherche mondiale en IA', 
                  fontsize=16, fontweight='bold', pad=20)
        plt.xlabel('Année')
        plt.ylabel('Publications (Milliers)')
        
        # Annotation
        plt.annotate('Mobilisation massive\ndes ressources mondiales', 
                     xy=(2023, 240), xytext=(2018, 200),
                     arrowprops=dict(facecolor='purple', shrink=0.05), color='purple', fontweight='bold', fontsize=12)

        plt.tight_layout()
        plt.show()

    # --- TEXTE D'ANALYSE 1 ---
    analyse_1 = """
### 1. L'Origine : Le Tsunami Scientifique
Tout part d'une effervescence mondiale. Contrairement à d'autres bulles technologiques, l'IA mobilise la totalité de la communauté scientifique informatique.
* **Le constat :** Le nombre de publications a doublé en moins de 5 ans.
* **La conséquence :** Chaque papier publié correspond à des centaines d'heures d'entraînement de modèles, créant une demande de base insatiable.
    """
    display(Markdown(analyse_1))

    print("Génération du Graphique 2 (Compute)...")

    # --- GRAPHIQUE 2 : LA SOIF DE CALCUL ---
    path_compute = 'data/Data2/Data/fig_1.3.17.csv'
    
    if os.path.exists(path_compute):
        try:
            df = pd.read_csv(path_compute)
            df['Date'] = pd.to_datetime(df['Publication date'])
            col_compute = [c for c in df.columns if 'compute' in c][0]
            
            plt.figure()
            sns.scatterplot(data=df, x='Date', y=col_compute, hue='Label', size=col_compute, sizes=(50, 500), legend=False)
            
            plt.title('LE MOTEUR DE LA CRISE : Explosion de la puissance de calcul requise (Échelle Log)', 
                      fontsize=16, fontweight='bold', pad=20)
            plt.xlabel('Date de Publication')
            plt.ylabel('Opérations de Calcul (PetaFLOPs) - Log Scale')
            plt.yscale('log')
            
            # Annotation
            plt.text(pd.Timestamp('2022-01-01'), 1e8, "Croissance Exponentielle\n(Demande Énergétique x1000)", 
                     color='red', fontweight='bold', ha='center', bbox=dict(facecolor='white', alpha=0.8))

            plt.tight_layout()
            plt.show()
        except Exception as e:
            print(f"Erreur graphique Compute: {e}")

    # --- TEXTE D'ANALYSE 2 ---
    analyse_2 = """
### 2. Le Moteur du Crash : L'Exponentielle Verticale
Le graphique ci-dessus est le plus inquiétant de tout le rapport.
* **L'échelle Logarithmique :** L'axe vertical est logarithmique. Une ligne qui monte ici signifie une explosion verticale dans la réalité.
* **Le constat :** La demande de calcul pour les IA de pointe double tous les quelques mois. C'est bien plus rapide que la Loi de Moore (progrès des puces).
* **La conséquence :** Pour combler l'écart, on multiplie simplement le nombre de machines (et donc la facture d'électricité).
    """
    display(Markdown(analyse_2))

    print("Génération du Graphique 3 (CO2)...")

    # --- GRAPHIQUE 3 : L'EMPREINTE CARBONE ---
    path_co2 = 'data/Data2/Data/fig_1.4.8.csv'
    
    if os.path.exists(path_co2):
        df = pd.read_csv(path_co2)
        df_subset = df[df['Model'].isin(['AlexNet', 'VGG16', 'BERT-Large'])]
        
        plt.figure()
        sns.barplot(data=df_subset, x='Model', y='CO2 equivalent emissions (tons)', palette='Reds')
        
        plt.title('IMPACT CLIMATIQUE : L\'évolution des émissions de CO2 par entraînement', 
                  fontsize=16, fontweight='bold', pad=20)
        plt.xlabel('Modèle (Année)')
        plt.ylabel('Émissions CO2 (Tonnes)')
        
        # Annotation sur BERT
        val_bert = df_subset.loc[df_subset['Model']=='BERT-Large', 'CO2 equivalent emissions (tons)'].values[0]
        plt.text(2, val_bert + 0.1, f"Explosion des émissions\n({val_bert} Tonnes)", 
                 ha='center', color='darkred', fontweight='bold')

        plt.tight_layout()
        plt.show()

    # --- TEXTE DE CONCLUSION ---
    conclusion = """
### 3. Verdict : Un Modèle Insoutenable
Le dernier graphique illustre le prix physique de cette course. Entre un modèle de 2012 et un modèle de 2018, la pollution a été multipliée par **200**.
Avec les modèles de 2024 (non affichés car hors échelle), nous atteignons des sommets qui rendent la neutralité carbone impossible sans changement radical de stratégie.
    """
    display(Markdown(conclusion))

if __name__ == "__main__":
    visualiser_explosion_ia_avec_texte()
```
