# Infrastructures : data centers, timelines et refroidissement

Panorama rapide des sites de data centers, de leur montée en puissance (MW) et des besoins de refroidissement.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)

root = Path('../data/data_centers')
centers = pd.read_csv(root / 'data_centers.csv')
timeline = pd.read_csv(root / 'data_center_timelines.csv')

# Puissance par data center
ax = sns.barplot(data=centers.sort_values('Power (MW)', ascending=False), x='Power (MW)', y='Title', palette='flare')
ax.set_title('Puissance nominale des data centers listés')
plt.tight_layout()
plt.show()
```

```python
# Chronologie des mises en service
timeline['Date'] = pd.to_datetime(timeline['Date'], errors='coerce')
timeline = timeline.dropna(subset=['Date'])
timeline = timeline.sort_values('Date')

ax = sns.lineplot(data=timeline, x='Date', y='Power (MW)', hue='Construction status', marker='o')
ax.set_title('Puissance annoncée au fil du temps')
plt.tight_layout()
plt.show()
```

```python
# Refroidissement : capacité des tours et chillers
cooling = pd.read_csv(root / 'data_center_cooling_towers.csv')
chillers = pd.read_csv(root / 'data_center_chillers.csv')

cooling['Cooling capacity (central estimate, kW)'] = pd.to_numeric(cooling['Cooling capacity (central estimate, kW)'], errors='coerce')
chillers['Cooling capacity (kW)'] = pd.to_numeric(chillers['Cooling capacity (kW)'], errors='coerce')

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(cooling['Cooling capacity (central estimate, kW)'].dropna(), bins=30, ax=axes[0], color='royalblue')
axes[0].set_title('Capacité tours de refroidissement (kW)')
sns.histplot(chillers['Cooling capacity (kW)'].dropna(), bins=30, ax=axes[1], color='seagreen')
axes[1].set_title('Capacité chillers (kW)')
plt.tight_layout()
plt.show()
```

## Lecture rapide
- Les plus gros sites dépassent déjà plusieurs centaines de MW : ils pèsent comme des centrales électriques locales.
- La chronologie montre une montée graduelle du parc ; les statuts de construction aident à suivre la capacité en pipeline.
- Les besoins de refroidissement (kW) montrent l’ampleur des auxiliaires électriques et hydrauliques, à intégrer dans les bilans énergie/eau.
