# Mix électrique mondial et capacités

Analyse rapide des capacités et de la production par technologie (scénarios courant vs stated), à relier à la croissance des usages IA.
Chemins de données : `../data/preproc/total_power_generation.xlsx` et `../data/preproc/total_power_capacity.xlsx`.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)

root = Path('..') / 'data' / 'preproc'
prod = pd.read_excel(root / 'total_power_generation.xlsx')
cap = pd.read_excel(root / 'total_power_capacity.xlsx')

# On passe en format long pour tracer par techno
prod_long = prod.melt(id_vars='Technology', var_name='horizon', value_name='TWh')
prod_long[['year','scenario']] = prod_long['horizon'].str.extract(r'(\d{4})_(.*)')

# Focus sur quelques horizons
focus_years = ['2024','2035','2050']
pivot = prod_long[prod_long['year'].isin(focus_years)]

ax = sns.barplot(data=pivot, x='year', y='TWh', hue='Technology', palette='tab20')
ax.set_title('Production électrique par technologie (TWh)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Capacités installées (GW équivalent)
cap_long = cap.melt(id_vars='Technology', var_name='horizon', value_name='GW')
cap_long[['year','scenario']] = cap_long['horizon'].str.extract(r'(\d{4})_(.*)')
cap_focus = cap_long[cap_long['year'].isin(['2024','2035','2050'])]

ax = sns.barplot(data=cap_focus, x='year', y='GW', hue='Technology', palette='Paired')
ax.set_title('Capacités électriques installées par technologie')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Lecture rapide
- Les scénarios montrent la montée des renouvelables dans la production et les capacités à horizon 2035-2050.
- Les technologies pilotables restent nécessaires : relier ces volumes à la demande IA/crypto croissante aide à dimensionner l’appoint.
- Ces ordres de grandeur servent à tester si l’“energy wall” est atteignable ou non selon le mix.
